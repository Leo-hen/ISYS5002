```python
#Challenge: The Final Modular Code
# Define the calculator functions
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

def modulus(a, b):
    return a % b

# Display the menu
print("Simple Calculator Menu")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")
print("5. Modulus")

# Get the user's choice
choice = input("Enter your choice (1-5): ")

# Get the two numbers from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# Perform the chosen operation
if choice == "1":
    result = add(num1, num2)
elif choice == "2":
    result = subtract(num1, num2)
elif choice == "3":
    result = multiply(num1, num2)
elif choice == "4":
    result = divide(num1, num2)
elif choice == "5":
    result = modulus(num1, num2)
else:
    result = "Invalid choice."

# Display the result
print("Result:", result)
```

```python
#Challenge: advanced calculations
# Define the calculator functions
import pyinputplus as pyip

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

def modulus(a, b):
  return a % b

# Display the menu
print("Simple Calculator Menu")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")
print("5. Modulus")

# Get the user's choice using pyinputplus for validation
# Using timeout and limit
choice = pyip.inputChoice(['1', '2', '3', '4', '5'], prompt="Enter your choice (1-5): ", timeout=30, limit=3)

# Get the first number from the user using pyinputplus for validation
# Ensuring the input is positive (absolute value) for num1
num1 = pyip.inputNum("Enter the first number (will be made positive): ", timeout=30, limit=3, applyFunc=abs)

# Get the second number from the user using pyinputplus for validation
# If dividing, prevent zero; otherwise, allow any number
if choice == "4":
    num2 = pyip.inputNum("Enter the second number (cannot be zero for division): ", timeout=30, limit=3, blockRegex=r'^0(\.0*)?$', ) # Block 0 or 0.0 etc.
else:
    num2 = pyip.inputNum("Enter the second number: ", timeout=30, limit=3)

# Perform the chosen operation
if choice == "1":
    result = add(num1, num2)
elif choice == "2":
    result = subtract(num1, num2)
elif choice == "3":
    result = multiply(num1, num2)
elif choice == "4":
    result = divide(num1, num2) # No need for if num2==0 check here, pyinputplus handles it
elif choice == "5":
    result = modulus(num1, num2)
else:
    result = "Invalid choice."

# Display the result
print("Result:", result)
```
