```python
#Challenge: Weekly Weather Report

# Ensure 'temperatures' list is available for this section
temperatures = [18, 22, 25, 19, 30, 27, 21]
conditions = ['sunny', 'sunny', 'cloudy', 'rain', 'sunny', 'windy', 'cloudy']

# Highest and lowest reading of the week
highest_reading = max(temperatures)
lowest_reading = min(temperatures)

print(f"Highest reading: {highest_reading}°C")
print(f"Lowest reading: {lowest_reading}°C")

# Average reading, rounded to one decimal place
average_reading = sum(temperatures) / len(temperatures)
rounded_average = round(average_reading, 1)

print(f"Average reading: {rounded_average}°C")

# How many days were above the average
days_above_average = 0
for temp in temperatures:
    if temp > average_reading:
        days_above_average += 1

print(f"{days_above_average} days were above the average of {rounded_average}°C")

# Message for each day using if-else logic

print("\n--- Daily Weather Advice ---")
for i, temp in enumerate(temperatures):
    day = i + 1
    condition = conditions[i]
    print(f"Day {day}: Temperature {temp}°C, Condition: {condition}.")

    if temp > 25 and condition == 'sunny':
        print("  It's a hot and sunny day, perfect for the beach!")
    elif temp > 20 and condition == 'sunny':
        print("  A warm and sunny day, ideal for outdoor activities.")
    elif temp > 20 and condition == 'cloudy':
        print("  Warm with clouds, good for a walk.")
    elif temp < 15 and condition == 'rain':
        print("  Cold and rainy, stay indoors with a hot drink.")
    elif condition == 'rain':
        print("  Don't forget your umbrella today!")
    elif condition == 'windy':
        print("  It's quite windy, hold onto your hats!")
    else:
        print("  Enjoy the day!")
```

```python
#xtension: Reuse the modular approach from Week 4.
def calculate_average(readings):
    if not readings:
        return 0
    return sum(readings) / len(readings)

def count_above_threshold(readings, threshold):
    count = 0
    for reading in readings:
        if reading > threshold:
            count += 1
    return count

# Main program using functions

temperatures = [18, 22, 25, 19, 30, 27, 21]
conditions = ['sunny', 'sunny', 'cloudy', 'rain', 'sunny', 'windy', 'cloudy']

print("\n--- Weekly Weather Report (Modular) ---")

highest = max(temperatures)
lowest = min(temperatures)
print(f"Highest reading: {highest}°C")
print(f"Lowest reading: {lowest}°C")

weekly_average = calculate_average(temperatures)
rounded_weekly_average = round(weekly_average, 1)
print(f"Average reading: {rounded_weekly_average}°C")

days_hotter_than_average = count_above_threshold(temperatures, weekly_average)
print(f"{days_hotter_than_average} days were above the average of {rounded_weekly_average}°C")

print("\n--- Daily Advice ---")
for i, temp in enumerate(temperatures):
    day = i + 1
    condition = conditions[i]
    print(f"Day {day}: Temperature {temp}°C, Condition: {condition}.")

    if temp > 25 and condition == 'sunny':
        print("  It's a hot and sunny day, perfect for the beach!")
    elif temp > 20 and condition == 'sunny':
        print("  A warm and sunny day, ideal for outdoor activities.")
    elif temp > 20 and condition == 'cloudy':
        print("  Warm with clouds, good for a walk.")
    elif temp < 15 and condition == 'rain':
        print("  Cold and rainy, stay indoors with a hot drink.")
    elif condition == 'rain':
        print("  Don't forget your umbrella today!")
    elif condition == 'windy':
        print("  It's quite windy, hold onto your hats!")
    else:
        print("  Enjoy the day!")
```

```python
#Challenge: The Complete Weather Station
weather_data = []

def display_menu():
    print("\n--- Weather Station Menu ---")
    print("1. Add a new reading")
    print("2. Display all readings")
    print("3. Show summary statistics")
    print("4. Get daily advice")
    print("5. Exit")
    print("--------------------------")

def add_reading(data_list):
    while True:
        try:
            temp_str = input("Enter temperature (°C): ")
            temperature = float(temp_str)
            break
        except ValueError:
            print("Invalid temperature. Please enter a number.")

    condition = input("Enter condition (e.g., sunny, cloudy, rain, windy): ").lower()
    data_list.append((temperature, condition))
    print(f"Reading added: {temperature}°C, {condition}")

def show_readings(data_list):
    if not data_list:
        print("No readings recorded yet.")
        return

    print("\n--- All Recorded Readings ---")
    for i, (temp, condition) in enumerate(data_list):
        print(f"{i + 1}. Temperature: {temp}°C, Condition: {condition}")

def calculate_average(readings_list):
    if not readings_list:
        return 0.0
    return sum(readings_list) / len(readings_list)

def count_above_threshold(readings_list, threshold):
    count = 0
    for reading in readings_list:
        if reading > threshold:
            count += 1
    return count

def show_statistics(data_list):
    if not data_list:
        print("No readings recorded yet to show statistics.")
        return

    temperatures_only = [item[0] for item in data_list]

    count = len(temperatures_only)
    highest = max(temperatures_only)
    lowest = min(temperatures_only)
    average = calculate_average(temperatures_only)
    rounded_average = round(average, 1)
    days_above_average = count_above_threshold(temperatures_only, average)

    print("\n--- Summary Statistics ---")
    print(f"Total readings: {count}")
    print(f"Highest reading: {highest}°C")
    print(f"Lowest reading: {lowest}°C")
    print(f"Average reading: {rounded_average}°C")
    print(f"{days_above_average} days were above the average of {rounded_average}°C")

def get_daily_advice(temp, condition):
    print(f"Day's Temperature: {temp}°C, Condition: {condition}.")
    if temp > 25 and condition == 'sunny':
        print("  It's a hot and sunny day, perfect for the beach!")
    elif temp > 20 and condition == 'sunny':
        print("  A warm and sunny day, ideal for outdoor activities.")
    elif temp > 20 and condition == 'cloudy':
        print("  Warm with clouds, good for a walk.")
    elif temp < 15 and condition == 'rain':
        print("  Cold and rainy, stay indoors with a hot drink.")
    elif condition == 'rain':
        print("  Don't forget your umbrella today!")
    elif condition == 'windy':
        print("  It's quite windy, hold onto your hats!")
    else:
        print("  Enjoy the day!")

# Main program loop
while True:
    display_menu()
    choice = input("Enter your choice (1-5): ")

    if choice == '1':
        add_reading(weather_data)
    elif choice == '2':
        show_readings(weather_data)
    elif choice == '3':
        show_statistics(weather_data)
    elif choice == '4':
        if not weather_data:
            print("No readings recorded yet to give advice.")
            continue
        while True:
            try:
                day_index_str = input(f"Enter day number (1-{len(weather_data)}) for advice: ")
                day_index = int(day_index_str) - 1
                if 0 <= day_index < len(weather_data):
                    temp, condition = weather_data[day_index]
                    get_daily_advice(temp, condition)
                    break
                else:
                    print(f"Invalid day number. Please enter a number between 1 and {len(weather_data)}.")
            except ValueError:
                print("Invalid input. Please enter a whole number.")
    elif choice == '5':
        print("Exiting Weather Station. Goodbye!")
        break
    else:
        print("Invalid choice. Please enter a number between 1 and 5.")
```
