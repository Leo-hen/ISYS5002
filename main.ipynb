```python
import csv


def show_menu():
    print("\n===== Meal & Muscle Gain Tracker =====")
    print("1. Add meal")
    print("2. View daily summary")
    print("3. Add body weight")
    print("4. View weight progress")
    print("5. Check daily goals")
    print("6. View weekly summary")
    print("0. Exit")


def add_meal():
    print("\n--- Add Meal ---")

    date = input("Enter date (YYYY-MM-DD): ")
    meal_name = input("Enter meal name: ")

    try:
        calories = float(input("Enter calories (kcal): "))
        protein = float(input("Enter protein (g): "))
        carbs = float(input("Enter carbohydrates (g): "))
        fat = float(input("Enter fat (g): "))

    except ValueError:
        print("Invalid input. Calories and nutrients must be numbers.")
        return

    if calories < 0 or protein < 0 or carbs < 0 or fat < 0:
        print("Values cannot be negative.")
        return

    with open("meals.csv", "a", newline="") as file:
        writer = csv.writer(file)

        writer.writerow([
            date,
            meal_name,
            calories,
            protein,
            carbs,
            fat
        ])

    print("\nMeal successfully saved!")

    print("\nMeal information:")
    print("Date:", date)
    print("Meal:", meal_name)
    print("Calories:", calories, "kcal")
    print("Protein:", protein, "g")
    print("Carbohydrates:", carbs, "g")
    print("Fat:", fat, "g")


def main():
    while True:

        show_menu()

        choice = input("\nChoose an option: ")

        if choice == "1":
            add_meal()

        elif choice == "2":
            print("\nDaily summary will be added later.")

        elif choice == "3":
            print("\nBody weight tracking will be added later.")

        elif choice == "4":
            print("\nWeight progress will be added later.")

        elif choice == "5":
            print("\nDaily goals will be added later.")

        elif choice == "6":
            print("\nWeekly summary will be added later.")

        elif choice == "0":
            print("\nThank you for using Meal & Muscle Gain Tracker.")
            print("Goodbye!")
            break

        else:
            print("\nInvalid option. Please choose from 0 to 6.")


main()
```
